### Data Cleaning

```python
df['milage']=(
    df['milage']
    .str.replace('mi.','')
    .str.replace(',','')
    .astype('float')
)
```

```python
df.head()
```

```python
df['milage'].dtypes
```

```python
df.head()
```

```python
df['price']=(
    df['price']
    .str.replace('$','')
    .str.replace(',','')
    .astype(float)
)
```

```python
df.head()
```

```python
df['Engine_Capacity']=(
    df['engine']
    .str.extract(r'(\d+\.\d+L|\d+L|\d+\.\d+\s+Liter|\d+\s+Liter)')
)
```

```python
df.head()
```

```python
df['Horse_Power']=(
    df['engine']
    .str.extract(r'(\d+\.\d+HP|\d+HP)')
)
```

```python
df.head()
```

```python
df['Engine_Capacity']=(
    df['Engine_Capacity']
    .str.replace('Liter','')
    .str.replace('L','')
    .astype(float)
)
```

```python
df['Horse_Power']=(
    df['Horse_Power']
    .str.replace('HP','')
    .astype(float)
)
```

```python
df.head()
```

```python
df['Accident_Occured']=(
    df['accident'].apply(lambda x:"NO" if x=="None reported" else "YES")
)
```

```python
df.drop('accident',axis=1,inplace=True)
```

### Finding Missing Percentage

```python
missing_info={}
for col in df.columns:
    if df[col].isnull().sum()>0:
        missing_count=(df[col].isnull().sum()/len(df))*100
        missing_info[col]=round(missing_count,2)
print("\n The columns with missing percentage is:\n")
for col, per in missing_info.items():
    print(f'{col}:{per}%')
```
